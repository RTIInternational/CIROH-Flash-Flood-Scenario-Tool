```python
"""
SAC-SMA parameters are calibrated as global scalars (one value applied uniformly
to all HRUs). The full model chain SAC-SMA → UHG is run each
evaluation, and the objective is defined by obj_func and calculated on hourly observed streamflow.

THIS IS AN ALTERNATIVE (IN NOTEBOOK) TO RUNNING run_sceua_sac.py COMMAND LINE

"""

from pathlib import Path
import os
import numpy as np
import pandas as pd
import spotpy
import pysacsnow as pss

from spot_setup_sac import spot_setup

import holoviews as hv
hv.extension('bokeh') 
from bokeh.models import DatetimeTickFormatter, MonthsTicker
```

```python
# List of usgs_ids (dynamically extracted from pyMARRMot_input folder)
import glob
usgs_files = glob.glob(r"..\pyMARRMot_input\usgs-*_daily_marrmot.csv")
usgs_ids = sorted(set([os.path.basename(f).split('_')[0].split("-")[1] for f in usgs_files]))  # 16103000 format
```

```python
for usgs_id in usgs_ids:
    # basin = "16103000"
    basin = usgs_id
    base_dir = Path('C:/Users/hlohman/OneDrive - Research Triangle Institute/Documents/Python Scripts/pysac/hawaii')
    start_date = '2021-04-20'
    end_date = '2026-04-12'

    obj_func = 'kge_q90'
    warmup_hrs = 4380         # 6 months due to short PoR

    output_dir = Path(base_dir, 'data', 'output', f'basin_{basin}')
    output_dir.mkdir(parents=True, exist_ok=True)

    obs_path = Path(base_dir, f'data/observations/QIN_{basin}.csv')
    sac_control_yaml = Path(base_dir, f'configurations/sac/sac.control.{basin}.yaml')
    uhg_control_yaml = Path(base_dir, f'configurations/unithg/uhg.control.{basin}.yaml')
    db_name = str(output_dir / f'sceua_params_{basin}_{obj_func}')

    # ------------------------------------------------------------------
    # Build setup
    # ------------------------------------------------------------------
    setup = spot_setup(
        basin                = basin,
        obs_path             = obs_path,
        sac_control_yaml     = sac_control_yaml,
        uhg_control_yaml     = uhg_control_yaml,
        start_date           = start_date,
        end_date             = end_date,
        warmup_hrs           = warmup_hrs,
        obj_func             = obj_func
    )

    n_params = len(setup._spotpy_params)
    print(f'Basin        : {basin}')
    print(f'HRU IDs      : {setup._hru_ids}')
    print(f'Free params  : {n_params}  '
        f'({len(setup._CALIBRATED_PARAMS)} params x {len(setup._hru_ids)} zones)')

    # ------------------------------------------------------------------
    # Run SCEUA optimisation
    # parallel='mpi'  ->  mpirun -n N python run_optimization_sac_alec2.py
    # ------------------------------------------------------------------
    sampler = spotpy.algorithms.sceua(
        setup,
        dbname   = db_name,
        dbformat = 'csv',
        # parallel = 'mpi',
        save_sim = False,
    )

    reps = 10000
    ngs = 50
    kstop = 15
    pcento = 0.01

    sampler.sample(
        repetitions = reps,
        ngs         = ngs,
        kstop       = kstop,
        peps        = pcento,
    )
    print (f'Finished SCEUA optimization for basin {basin} with objective function {obj_func}. Results saved to {db_name}.csv')
```
