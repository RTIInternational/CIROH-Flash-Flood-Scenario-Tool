```python
"""
Run the optimized/best SAC-SMA → UHG simulations and plot with observed
"""

from pathlib import Path
import os
import numpy as np
import pandas as pd
import pysacsnow as pss
import spotpy
import json

import holoviews as hv
hv.extension('bokeh') 
from bokeh.models import DatetimeTickFormatter, MonthsTicker

base_dir = Path('C:/Users/hlohman/OneDrive - Research Triangle Institute/Documents/Python Scripts/pysac/hawaii')
```

```python
# List of usgs_ids (dynamically extracted from pyMARRMot_input folder)
import glob
usgs_files = glob.glob(r"..\pyMARRMot_input\usgs-*_daily_marrmot.csv")
usgs_ids = sorted(set([os.path.basename(f).split('_')[0].split("-")[1] for f in usgs_files]))  # 16103000 format
```

```python
def runsac_from_t0(sac_control, sacsma, uhg):
    # reset the models to t0
    sacsma.reset()
    uhg.reset()

    # Run models
    for tt in range(sac_control.n_times):
        sac_control.advance()
        sacsma.advance()
        sacsma.calculate()
        sacsma.output()
    
        uhg_control.advance()
        uhg.advance()
        uhg.hru_tci[:] = sacsma.tci[:]
        uhg.calculate()
        uhg.output()
    
    basin_tci = sacsma.buffer_states["tci"]
    total_flow = np.sum(uhg.buffer_states["q"], axis=1)
    flow_df = pd.DataFrame({"flow_sim": total_flow}, index=uhg.value_time_state)
    
    return flow_df
```

```python
# Get best parameters for each basin and save to csv
for usgs_id in usgs_ids:
    basin = usgs_id
    start_date = '2021-04-20'
    end_date = '2026-04-12'
    obj_func = 'kge_q90'

    output_dir = Path(base_dir, 'data', 'output', f'basin_{basin}')
    spotpy_output_path = Path(output_dir, f'sceua_params_{basin}_{obj_func}.csv')
    spotpy_db_name = Path(spotpy_output_path.parent, spotpy_output_path.name.replace('.csv',''))
    obs_path = Path(base_dir, f'data/observations/QIN_{basin}.csv')

    sac_control_yaml = Path(base_dir, f'configurations/sac/sac.control.{basin}.yaml')
    uhg_control_yaml = Path(base_dir, f'configurations/unithg/uhg.control.{basin}.yaml')
    sac_control = pss.Controller.load_from_yaml(sac_control_yaml)
    uhg_control = pss.Controller.load_from_yaml(uhg_control_yaml )
    sacsma = pss.SACSMA(control=sac_control, calc_method="numpy")
    uhg = pss.UHG(control=uhg_control, calc_method="numpy")

    if spotpy_output_path.exists():
        print('Reading ', spotpy_output_path)
        results = spotpy.analyser.load_csv_results(str(spotpy_db_name))

    #  -- best = minimum likelihood value
    best_row  = results[np.argmin(results['like1'])]
    best_obj  = float(best_row['like1'])
    best_nkge = 1.0 - best_obj
    best_kge  = 2.0 * best_nkge / (1.0 + best_nkge)

    calibrated_params = [
        name.split("_")[0].replace("par", "")
        for name in results.dtype.names
        if name.startswith("par")
    ]
    hru_ids = sacsma.params.spatial_ids

    # Extract best parameters -- one value per (param, zone) pair
    best_params: dict[str, dict[str, float]] = {}
    for param in calibrated_params:
        best_params[param] = {
            zone: float(best_row[f'par{param}_{zone}'])
            for zone in hru_ids
        }

    # Print summary
    print(f'Reading  {spotpy_output_path}')
    print('\n' + '=' * 60)
    print(f'Calibration results  {basin}')
    print('=' * 60)
    print(f'Total evaluations : {len(results)}')
    print(f"Best NKGE'        : {best_nkge:.4f}  "
        f"(1-NKGE' = {best_obj:.4f})")
    print(f"Best KGE'         : {best_kge:.4f}")
    print(f'\nBest parameter set:')
    header = f"  {'Parameter':<10s}" + ''.join(f'  {z:>14s}' for z in hru_ids)
    print(header)
    print('  ' + '-' * (len(header) - 2))

    for param in calibrated_params:
        row = f'  {param:<10s}'
        for zone in hru_ids:
            row += f'  {best_params[param][zone]:>14.6f}'
        print(row)

    # Save best params as tidy CSV: columns = [parameter, zone, best_value]
    rows = [
        {'parameter': param, 'zone': zone, 'best_value': val}
        for param, zone_vals in best_params.items()
        for zone, val in zone_vals.items()
    ]
    best_path = output_dir / f'best_params_{basin}_{obj_func}.csv'
    pd.DataFrame(rows).to_csv(best_path, index=False)
    print(f'\nBest params saved -> {best_path}')

    sacsma = pss.SACSMA(control=sac_control, calc_method="numpy")
    uhg = pss.UHG(control=uhg_control, calc_method="numpy")

    # check params
    for param in calibrated_params:
        print(f'{param}: default value: {getattr(sacsma, param)}')

    # run the model
    default_df = runsac_from_t0(sac_control, sacsma, uhg)

    # apply the best parameters
    for param in calibrated_params:
        per_hru_dict = best_params[param]
        per_hru = np.array(
            [float(per_hru_dict[key]) for key in per_hru_dict.keys()]
        )    
        setattr(sacsma, param, per_hru)

    # check they've been set
    for param in calibrated_params:
        print(f'{param}: calibrated value: {getattr(sacsma, param)}')

    calb_df = runsac_from_t0(sac_control, sacsma, uhg)

    obs_df = pd.read_csv(obs_path, parse_dates=['time'], index_col='time')
    obs_df = obs_df.loc[default_df.index.min(): default_df.index.max()]
    merge_df = pd.concat([default_df, obs_df], axis=1).reset_index().rename(columns={'index':'time'})
    merge_calb_df = pd.concat([calb_df, obs_df], axis=1).reset_index().rename(columns={'index':'time'})

    opts = dict(
        width=1100,
        height=400,
        xlabel='Datetime',
        ylabel='cms',
        tools=['hover'],
        xrotation=90,
        active_tools=['reset'],
        toolbar='above'
    )

    sim = hv.Curve(merge_df, kdims='time', vdims='flow_sim', label='sim_default').opts(**opts)
    sim_calb = hv.Curve(merge_calb_df, kdims='time', vdims='flow_sim', label='sim_calb').opts(**opts)
    obs = hv.Curve(merge_df, kdims='time', vdims='QIN_cms', label='obs').opts(**opts, line_color = 'black', line_dash = 'dashed')
    (sim * sim_calb * obs).opts(legend_position='right')

    chart = (sim * sim_calb * obs).opts(legend_position='right')
    file_name = Path(output_dir) / f"calibration_results_plot_{basin}.html"
    hv.save(chart, str(file_name))
```

```python
# Create configuration files for calibrated parameters
for usgs_id in usgs_ids:
    basin = usgs_id
    output_dir = Path(base_dir, 'data', 'output', f'basin_{basin}')
    best_path = output_dir / f'best_params_{basin}_kge_q90.csv'
    sac_config_path = Path(base_dir, 'configurations', 'sac')
    sac_params_path = sac_config_path / f'sac_params_{basin}.json'
    sac_params_cal_path = sac_config_path / f'sac_params_{basin}_cal.json'

    # Load best parameters
    best_params_df = pd.read_csv(best_path)

    # Load default configuation file
    with open(sac_params_path, 'r') as f:
        sac_json_content = json.load(f)

    # Update the JSON content with best parameters
    for _, row in best_params_df.iterrows():
        param = row['parameter']
        best_value = row['best_value']

        # Assuming the JSON structure has a 'parameters' section with zones
        if param in sac_json_content.keys():
            sac_json_content[param] = [best_value]
        else:
            print(f"Warning: Parameter '{param}' not found in JSON")

    # Save the updated JSON content to a new file
    with open(sac_params_cal_path, 'w') as f:
        json.dump(sac_json_content, f, indent=4)
```
