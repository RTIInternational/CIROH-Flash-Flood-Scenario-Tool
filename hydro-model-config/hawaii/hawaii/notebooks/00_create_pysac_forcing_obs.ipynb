```python
usgs_ids = {
    f.name.split("_")[0]
    for f in read_dir.glob("usgs-*.parquet")
}
usgs_ids = sorted(usgs_ids)
```

```python
usgs_ids = sorted(usgs_ids)
df = pd.read_parquet(Path(base_dir, 'usgs_point_attr_drainage_area.all.parquet'))
df_areas = df[df['location_id'].isin(usgs_ids)]

area_km2 = df_areas.loc[df_areas['location_id'] == location_id, 'value'].item()
```

```python
for usgs_id in usgs_ids:

    usgs_num = usgs_id.split("-")[1]
    filename = f'{usgs_id}_combined_extana.parquet'
    df = pd.read_parquet(Path(read_dir, filename))
    df["value_time"] = pd.to_datetime(df["value_time"])
    df = df.set_index("value_time")
    
    # Create full hourly time range
    full_index = pd.date_range(
        start=df.index.min(),
        end=df.index.max(),
        freq="h"
    )
    # Reindex to full range (this inserts missing timestamps)
    df = df.reindex(full_index)
    # Fill missing values by column
    df["precip_mm"] = df["precip_mm"].fillna(0)
    df["pet_mm"] = df["pet_mm"].fillna(0)
    # If you want datetime back as a column
    df = df.reset_index().rename(columns={"index": "time"})
    
    df_sac = df[['time','precip_mm','pet_mm']].rename(columns={'precip_mm':'hru_ppt','pet_mm':'hru_pet'})
    
    # round data to 3 decimals
    df_sac['hru_ppt'] = df_sac['hru_ppt'].round(4)
    df_sac['hru_pet'] = df_sac['hru_pet'].round(4)
    df_sac.to_csv(Path(forcing_dir, f'forcing.sac.{usgs_num}.csv'))

    # create uh forcing (all zero)
    df_uhg = df_sac.copy().rename(columns={'hru_ppt':'hru_tci'}).drop('hru_pet', axis=1)
    df_uhg['hru_tci'] = 0
    df_uhg.to_csv(Path(forcing_dir, f'forcing.uhg.{usgs_num}.csv'))

    # get area
    area_km2 = df_areas.loc[df_areas['location_id'] == usgs_id, 'value'].item()
    
    # create obs 
    df_obs = df[['time','obs_flow_mm']].rename(columns={'obs_flow_mm':'QIN_cms'})
    df_obs['QIN_cms'] = df_obs['QIN_cms'] * area_km2 * 1000 / 3600
    df_obs = df_obs.set_index('time')
    df_obs.to_csv(Path(obs_dir, f'QIN_{usgs_num}.csv'))

    print(usgs_id, area_km2)
```
