```python
"""
Single run of SAC-SMA → UHG using parameters in the defined json file
"""
from pathlib import Path
import numpy as np
import pandas as pd
import pysacsnow as pss
import json

import holoviews as hv
hv.extension('bokeh') 
from bokeh.models import DatetimeTickFormatter, MonthsTicker
```

```python
def read_stored_states(stored_states_path):
    # Read existing states JSON if it exists
    if stored_states_path.exists():
        with open(stored_states_path, "r") as f:
            stored_states = json.load(f)
    else:
        stored_states = {}
    return stored_states

def find_closest_states_date(start_date, stored_states):
    date_matches = []
    for date_str in stored_states.keys():
        date_np = np.datetime64(date_str)
        diff = abs(date_np - start_date)
        date_matches.append((diff, date_str))
    
    # Closest date
    _, nearest_date_str = min(date_matches, key=lambda x: x[0])
    print("Using states:", nearest_date_str)
    stored_states_date = stored_states[nearest_date_str]
    
    return stored_states_date

def runsac_from_t0(sac_control, sacsma, uhg):

    # Run models
    for tt in range(sac_control.n_times):
        sac_control.advance()
        sacsma.advance()
        sacsma.calculate()
        sacsma.output()
    
        uhg_control.advance()
        uhg.advance()
        uhg.hru_tci[:] = sacsma.tci[:]
        uhg.calculate()
        uhg.output()
    
    basin_tci = sacsma.buffer_states["tci"]
    total_flow = np.sum(uhg.buffer_states["q"], axis=1)
    flow_df = pd.DataFrame({"flow_sim": total_flow}, index=uhg.value_time_state)
    
    return flow_df

def store_latest_states(stored_states_path, sacsma):

    # Read existing states JSON if it exists
    if stored_states_path.exists():
        with open(stored_states_path, "r") as f:
            stored_states = json.load(f)
    else:
        stored_states = {}
    
    states = sacsma.get_variables()
    
    latest_date = sacsma.value_time_state.max()
    date_key = np.datetime_as_string(latest_date, unit='s')
    
    latest_states = {}
    for state in states:
        state_values = sacsma.buffer_states[state]
        state_values_latest_list = [float(value) for value in state_values[-1]]
        latest_states[state] = state_values_latest_list 
    
    stored_states[date_key] = latest_states

    # Write to JSON file
    with open(stored_states_path, "w") as f:
        json.dump(stored_states, f, indent=4)
    
    print("States file written: ", stored_states_path)
```

```python
base_dir = Path('/home/kvanwerkhoven/projects/pysac/hawaii')

basin = "16103000"
start_date = '2021-04-20'
end_date = '2026-04-12'

output_dir = Path(base_dir, 'data', 'output', f'basin_{basin}')
output_dir.mkdir(parents=True, exist_ok=True)

obs_path = Path(base_dir, f'data/observations/QIN_{basin}.csv')
sac_control_yaml = Path(base_dir, f'configurations/sac/sac.control.{basin}.yaml')
uhg_control_yaml = Path(base_dir, f'configurations/unithg/uhg.control.{basin}.yaml')
```

```python
sac_control = pss.Controller.load_from_yaml(sac_control_yaml)
uhg_control = pss.Controller.load_from_yaml(uhg_control_yaml )

###  to update the output or parameter filepaths (if desired, e.g., optimized parameters) 
sac_control.config.simulation.output_path = Path('../data/output/basin_16103000/output_sac_16103000_default.csv')
uhg_control.config.simulation.output_path = Path('../data/output/basin_16103000/output_uhg_16103000_default.csv')
stored_states_path = Path('../data/output/basin_16103000/states_sac_16103000.json')  # will need to update this with dates

# sac_control.config.model.parameters.path = Path('../configurations/sac/sac_params_16103000_opt1.json')
```

```python
# initialze the models
sacsma = pss.SACSMA(control=sac_control, calc_method="numpy")
uhg = pss.UHG(control=uhg_control, calc_method="numpy")

# reset the models
sacsma.reset()
uhg.reset()
```

```python
# read initial states
stored_states = read_stored_states(stored_states_path)

# find and set states for nearest date to start_date
stored_states_date = find_closest_states_date(sac_control.config.simulation.start_date, stored_states)
if stored_states_date:
    for state_name, state_value in stored_states_date.items():
        sacsma[state_name][:] = state_value
```

```python
# run the model
flow_df = runsac_from_t0(sac_control, sacsma, uhg)

# save model outputs
sacsma.finalize()
uhg.finalize()

# store latest states
store_latest_states(stored_states_path, sacsma)
```

```python
# read the observed
obs_df = pd.read_csv(obs_path, parse_dates=['time'], index_col='time')
obs_df = obs_df.loc[flow_df.index.min(): flow_df.index.max()]
merge_df = pd.concat([flow_df, obs_df], axis=1).reset_index().rename(columns={'index':'time'})

# plot the sim and obs
opts = dict(
    width=1100,
    height=400,
    xlabel='Datetime',
    ylabel='cms',
    tools=['hover'],
    xrotation=90,
    active_tools=['reset'],
    toolbar='above'
)
sim = hv.Curve(merge_df, kdims='time', vdims='flow_sim', label='sim_default').opts(**opts)
obs = hv.Curve(merge_df, kdims='time', vdims='QIN_cms', label='obs').opts(**opts, line_color = 'black', line_dash = 'dashed')
(sim * obs).opts(legend_position='right')
```
