```python
usgs_ids = {
    f.name.split("_")[0]
    for f in read_dir.glob("usgs-*.parquet")
}
usgs_ids = sorted(usgs_ids)
df = pd.read_parquet(Path(base_dir, 'pysac/usgs_point_attr_drainage_area.all.parquet'))
df_areas = df[df['location_id'].isin(usgs_ids)]

location_id = 'usgs-16103000'
area_km2 = df_areas.loc[df_areas['location_id'] == location_id, 'value'].item()
area_km2
```

```python
import numpy as np
import pandas as pd
from scipy.stats import gamma

def gamma_uh_instantaneous_cms_per_mm(
    area_km2,
    dt_hr=1,
    duration_hr=48,
    shape=2.5,
    scale=1.0,
):
    """
    Instantaneous unit hydrograph ordinates in cms for 1 mm runoff.

    Ordinates are point flows at t = 0, 1, 2, ... hours.
    The hydrograph volume integrates to:
        1 mm over area_km2 = area_km2 * 1000 m3
    """

    time_hr = np.arange(0, duration_hr + dt_hr, dt_hr)

    # Point values of gamma curve
    q_shape = gamma.pdf(time_hr, a=shape, scale=scale)

    # No instantaneous flow at t=0
    q_shape[0] = 0.0

    # Normalize so integral over time equals 1
    integral_hr = np.trapezoid(q_shape, time_hr)
    q_density_per_hr = q_shape / integral_hr

    # Convert from fraction-per-hour to cms
    total_volume_m3 = area_km2 * 1000.0
    uh_cms = q_density_per_hr * total_volume_m3 / 3600.0

    return pd.DataFrame({
        "time_hr": time_hr,
        "uh_cms": uh_cms
    })
```

```python
parameter_sets = [
    {"shape": 2.0, "scale": 0.75},
    {"shape": 2.5, "scale": 0.75},
    {"shape": 2.5, "scale": 1.0},
    {"shape": 3.0, "scale": 1.0},
]
```

```python
curves = []
uhs = []

for params in parameter_sets:
    shape = params['shape']
    scale = params['scale']
    label = f"shape={shape}, scale={scale}"
    
    uh = gamma_uh_instantaneous_cms_per_mm(
        area_km2=area_km2,
        dt_hr=1,
        duration_hr=48,
        shape=shape,
        scale=scale,
    )
    uh["trend"] = np.where(
        uh["uh_cms"].diff() > 0, "rising",
        np.where(uh["uh_cms"].diff() < 0, "falling", "flat")
    )
    keep_uh = uh[(uh['uh_cms'] > 0.01) | (uh['trend'] != 'falling')]
    uhs.append(keep_uh)
    
    curve = hv.Curve(uh, kdims='time_hr', vdims='uh_cms', label=label).opts(
        width=900,
        height=400,
        xlabel='hour',
        ylabel='cms',
        tools=['hover'],
        xlim = (-1,10)
    )
    curves.append(curve)
    
    volume_m3 = np.trapezoid(
        uh["uh_cms"],
        uh["time_hr"] * 3600
    )
    expected_m3 = area_km2 * 1000.0
    print(volume_m3, expected_m3)
```
